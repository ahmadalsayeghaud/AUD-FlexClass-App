```python
# Cell 5: Download and organise Kaggle datasets
print("\n" + "="*60)
print("DOWNLOADING KAGGLE DATASETS")
print("="*60)

kaggle_stats = []

for ds_id, friendly_name, subfolder in kaggle_datasets:
    print(f"\n📦 Processing: {ds_id} -> {friendly_name}")
    target_dir = os.path.join(BASE_DIR, friendly_name)

    # Skip if already downloaded (simple check)
    if os.path.exists(target_dir) and len(os.listdir(target_dir)) > 0:
        print(f"   ✅ Already exists, skipping download.")
        kaggle_stats.append((friendly_name, "already present"))
        continue

    try:
        # Download using kagglehub
        download_path = kagglehub.dataset_download(ds_id)
        print(f"   Downloaded to temporary location: {download_path}")

        # Move to our organised folder
        if subfolder:
            # If dataset has a specific subfolder we want, copy that
            source = os.path.join(download_path, subfolder)
            if os.path.exists(source):
                shutil.copytree(source, target_dir, dirs_exist_ok=True)
                print(f"   Copied subfolder '{subfolder}' to {target_dir}")
            else:
                # fallback: copy everything
                shutil.copytree(download_path, target_dir, dirs_exist_ok=True)
                print(f"   Copied all contents to {target_dir}")
        else:
            shutil.copytree(download_path, target_dir, dirs_exist_ok=True)
            print(f"   Copied all contents to {target_dir}")

        kaggle_stats.append((friendly_name, "downloaded"))
    except Exception as e:
        print(f"   ❌ Error: {e}")
        kaggle_stats.append((friendly_name, f"error: {e}"))

print("\n" + "="*60)
print("KAGGLE DOWNLOAD SUMMARY")
for name, status in kaggle_stats:
    print(f"{name}: {status}")
```
